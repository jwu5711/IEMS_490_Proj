```python
import h5py
import matplotlib.pyplot as plt
from pathlib import Path
import numpy as np
import pandas as pd
```

```python
folder_path = './DATASET' # Replace with your folder path
# This creates a list of extensions (including the dot)
file_paths = [f for f in Path(folder_path).iterdir() if f.is_file() and f.suffix == '.hdf5']

df = None
data = []
for idx, file in enumerate(file_paths):
    layer_dict = {}
    file_info = h5py.File(file, 'r')
    file_data = file_info['OpenData']
    x = file_data[0]
    y = file_data[1]
    p = file_data[5]
    ir = file_data[6]
    c1 = file_data[9]
    c2 = file_data[10]
    id1 = file_data[7]
    id2 = file_data[8]

    # print(file, np.unique(id1, return_counts=True), np.unique(id2, return_counts=True))
    
    layer_dict["file"] = file
    layer_dict["num_points"] = len(x)
    layer_dict["bulk_indicator"] = True if int(id1[0])== 1 else False
    layer_dict["first_3_post_anomaly"] = True if int(id2[0])!= 0 else False
    layer_dict["min_x"] = x.min()
    layer_dict["max_x"] = x.max()
    layer_dict["min_y"] = y.min()
    layer_dict["max_y"] = y.max()
    layer_dict["bulk_id"] = id1[0]
    layer_dict["ooc_id"] = id2[0]
    layer_dict["xa"] = -1
    layer_dict["xb"] = -1
    layer_dict["xc"] = -1
    layer_dict["xd"] = -1
    layer_dict["ya"] = -1
    layer_dict["yb"] = -1
    layer_dict["yc"] = -1
    layer_dict["yd"] = -1

    # x_mid = (x.min() + x.max())/2
    # y_mid = (y.min() + y.max())/2

    if len(x) < 10000:
        # print(file)
        ys_for_center_xs = y[(-26 < x) & (x < -20)]
        mid_y = (ys_for_center_xs.max() + ys_for_center_xs.min())/2
        # print("mid_y", mid_y)
        top_for_center_xs = ys_for_center_xs[ys_for_center_xs > mid_y] 
        ya = top_for_center_xs.max()
        yb = top_for_center_xs.min()
        bottom_for_center_xs = ys_for_center_xs[ys_for_center_xs < mid_y]
        yc = bottom_for_center_xs.max()
        yd = bottom_for_center_xs.min()
        
        xs_for_center_ys = x[(-88 < y) & (y < -82)]
        mid_x = (xs_for_center_ys.max() + xs_for_center_ys.min())/2
        # print("mid_x", mid_x)
        left_for_center_ys = xs_for_center_ys[xs_for_center_ys < mid_x]
        xa = left_for_center_ys.min()
        xb = left_for_center_ys.max()
        right_for_center_ys = xs_for_center_ys[xs_for_center_ys > mid_x]
        xc = right_for_center_ys.min()
        xd = right_for_center_ys.max()

        # print("xa", xa)
        # print("xb", xb)
        # print("xc", xc)
        # print("xd", xd)
        # print()
        # print("ya", ya)
        # print("yb", yb)
        # print("yc", yc)
        # print("yd", yd)
        # break

        layer_dict["xa"] = xa
        layer_dict["xb"] = xb
        layer_dict["xc"] = xc
        layer_dict["xd"] = xd
        layer_dict["ya"] = ya
        layer_dict["yb"] = yb
        layer_dict["yc"] = yc
        layer_dict["yd"] = yd

    

    if not isinstance(df, pd.DataFrame):
        df = pd.DataFrame(layer_dict, index=[idx])
    else:
        df = pd.concat([df, pd.DataFrame(layer_dict, index=[idx])])
```
